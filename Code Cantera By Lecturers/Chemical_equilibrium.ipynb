```python
import cantera as ct
```

```python
g = ct.Solution('gri30.yaml')
g.TPX = 300.0, ct.one_atm, 'CH4:0.95,O2:2,N2:7.52'
g.equilibrate('HP') #  fixed specific enthalpy and pressure
```

```python
rf = g.forward_rates_of_progress
rr = g.reverse_rates_of_progress

deviations = []

for i in range(g.n_reactions):

    # Only evaluate reversible reactions with nonzero forward rate
    if g.reaction(i).reversible and rf[i] != 0.0:

        # Relative net rate of progress:
        # Near equilibrium (rf ≈ rr)
        ratio = (rf[i] - rr[i]) / rf[i]

        deviations.append((abs(ratio), i))

        print(
            f"{i:4d}  {ratio:10.4g}   "
            f"{g.reaction(i).equation}"
        )

# Reaction farthest from equilibrium
max_dev, max_rxn = max(deviations)

print("\nMaximum deviation from equilibrium:")
print(f"Reaction index : {max_rxn}")
print(f"Deviation      : {max_dev:.4g}")
print(f"Reaction       : {g.reaction(max_rxn).equation}")
                                    
```

```python
g.equilibrate('UV') #  fixed specific internal energy and specific volume
g.equilibrate('SV') #  fixed specific entropy and specific volume
g.equilibrate('SP') #  fixed specific entropy and pressure
g.equilibrate('TP') #  fixed temperature and pressure
```
