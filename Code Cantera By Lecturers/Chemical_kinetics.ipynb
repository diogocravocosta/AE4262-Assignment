```python
import cantera as ct
```

```python
g = ct.Solution('gri30.yaml')       # Load GRI-Mech 3.0 mechanism
```

```python
r = g.reaction(2)                   # Get the 3rd reaction object
r.reactants                         # Reactant species and coefficients
r.products                          # Product species and coefficients
r.rate.input_data                   # Rate parameters (A, b, Ea)
r.equation                          # Reaction as a string equation
r.reversible                        # True if the reaction is reversible
r.duplicate                         # True if it's a duplicate reaction
r.ID                                # Reaction ID string
r.rate                              # Rate object (Arrhenius, etc.)
r.rate.pre_exponential_factor       # A in Arrhenius form
r.rate.temperature_exponent         # b in Arrhenius form
r.rate.activation_energy            # Ea in Arrhenius form

attrs = {
    "reactants": r.reactants,
    "products": r.products,
    "rate_input_data": r.rate.input_data,
    "equation": r.equation,
    "reversible": r.reversible,
    "duplicate": r.duplicate,
    "ID": r.ID,
    "rate": r.rate,
    "A": r.rate.pre_exponential_factor,
    "b": r.rate.temperature_exponent,
    "Ea": r.rate.activation_energy,
}

for k, v in attrs.items():
    print(f"{k}: {v}")
```
